```python
# =========================================================
# Estudo do Campeonato Brasileiro
# Distribuição de gols por período da partida
# =========================================================

query = """
WITH gols_por_periodo AS (
    SELECT
        CASE
            WHEN minuto_int BETWEEN 0 AND 15 THEN '00-15'
            WHEN minuto_int BETWEEN 16 AND 30 THEN '16-30'
            WHEN minuto_int BETWEEN 31 AND 45 THEN '31-45'
            WHEN minuto_int BETWEEN 46 AND 60 THEN '46-60'
            WHEN minuto_int BETWEEN 61 AND 75 THEN '61-75'
            WHEN minuto_int BETWEEN 76 AND 90 THEN '76-90'
            WHEN minuto_int BETWEEN 91 AND 105 THEN '91-105'
            WHEN minuto_int BETWEEN 106 AND 120 THEN '106-120'
            ELSE 'OUTRO'
        END AS periodo
    FROM fato_gol
)
SELECT
    periodo,
    COUNT(*) AS total_gols
FROM gols_por_periodo
GROUP BY periodo
ORDER BY total_gols DESC
"""

# Executa a consulta SQL
df_resultado = spark.sql(query)

# Exibe o resultado no notebook
display(df_resultado)
```
