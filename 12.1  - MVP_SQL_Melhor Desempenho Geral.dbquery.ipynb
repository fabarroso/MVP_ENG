```sql
WITH resultados AS (

    -- Mandante
    SELECT 
        p.partida_id,
        p.mandante AS clube,
        p.mandante_placar AS g_for,
        p.visitante_placar AS g_against
    FROM fato_partida p

    UNION ALL

    -- Visitante
    SELECT
        p.partida_id,
        p.visitante AS clube,
        p.visitante_placar AS g_for,
        p.mandante_placar AS g_against
    FROM fato_partida p
),

metricas AS (
    SELECT
        clube,
        COUNT(*) AS jogos,
        SUM(CASE WHEN g_for > g_against THEN 1 ELSE 0 END) AS vitorias,
        SUM(CASE WHEN g_for = g_against THEN 1 ELSE 0 END) AS empates,
        SUM(CASE WHEN g_for < g_against THEN 1 ELSE 0 END) AS derrotas,
        SUM(g_for - g_against) AS saldo_gols,
        SUM(
            CASE 
                WHEN g_for > g_against THEN 3 
                WHEN g_for = g_against THEN 1 
                ELSE 0 
            END
        ) AS pontos
    FROM resultados
    GROUP BY clube
)

SELECT *
FROM metricas
ORDER BY pontos DESC, saldo_gols DESC;
```
